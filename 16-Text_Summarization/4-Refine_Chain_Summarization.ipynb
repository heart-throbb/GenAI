### Refine Chain For Summarization
- Previous chunk Summarization result is combined with next chunk and is passed to prompt and llm  

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

api_key=os.getenv("GROQ_API_KEY")
llm=ChatGroq(groq_api_key=api_key,model="openai/gpt-oss-120b")
llm
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("I Have a Dream by Martin Luther King Jr.pdf")
docs = loader.load_and_split()
docs
```

```python
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter

final_documents=RecursiveCharacterTextSplitter(chunk_size=2000,chunk_overlap=100).split_documents(docs)
final_documents
```

```python
from langchain_classic.chains.summarize import load_summarize_chain

chain=load_summarize_chain(
    llm=llm,
    chain_type="refine",
    verbose=True
)
output_summary=chain.run(final_documents)
output_summary
```
