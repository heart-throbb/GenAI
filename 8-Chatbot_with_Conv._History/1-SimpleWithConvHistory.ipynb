```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key=os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model=ChatGroq(model="openai/gpt-oss-20b",groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content="Hi , My name is Anas and I am a software engineer and I am learning GenAI")])
```

```python
from langchain_core.messages import AIMessage
model.invoke(
    [
        HumanMessage(content="Hi , My name is Anas and I am a software engineer and I am learning GenAI"),
        AIMessage(content="Hello Anas! 👋 Great to meet a fellow software engineer diving into Generative AI. How can I help you today? Are you looking for resources, project ideas, code snippets, or something else? Let me know what's on your mind!"),
        HumanMessage(content="Hey What's my name and what I am learning?")
    ]
)
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store={}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history=RunnableWithMessageHistory(model,get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="Hi , My name is Anas and I am a software engineer and I am learning GenAI")],
    config=config
)
```

```python
response.content
```

```python
with_message_history.invoke(
    [HumanMessage(content="Whats my name?")],
    config=config
)
```

```python
## change the config-->session id
config1={"configurable":{"session_id":"chat2"}}
response=with_message_history.invoke(
    [HumanMessage(content="Whats my name")],
    config=config1
)
response.content
```
